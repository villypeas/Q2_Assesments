```python
def add(num1, num2):
  return num1 + num2
def subtract(num1, num2):
  return num1 - num2
def multiply(num1, num2):
  return num1 * num2
def divide(num1,num2):
  return "Number invalid zero cant be divided." if num2 == 0 else num1 / num2


num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Choose your operation of choice")
print("1 = Addition")
print("2 = Subtraction")
print("3 = Multiplication")
print("4 = Division")
operation_of_choice = int(input("What is your operation of choice?: "))

if operation_of_choice == 1:
  print("result", add(num1, num2))
elif operation_of_choice == 2:
  print("result", subtract(num1, num2))
elif operation_of_choice == 3:
  print("result", multiply(num1, num2))
elif operation_of_choice == 4:
  print("result", divide(num1, num2))
else :
  print("Invalid choice.")



  #Reflection
  #1. What functions did you create in your program?
    #add, subtraction, multiply, and divide
  #2. What parameters did your functions use?
    #num1 and num2
  #3. What arguements were passed down when the functions were called?
    #The user's input
  #4. How did your program use the returned value
    #It printed the returned value
  #5. Why is it better to divide the calculator program into functions instead of writing everything in long block of code?
    #Because it makes it easier to fix if theres problems, and more organized.
```
